part03

```python

# 1. Create lists
squares = [1, 4, 9, 16, 25]
empty = []

print("Squares:", squares)
print("Type:", type(squares))
print("Empty list:", empty)

# 2. Membership testing
print("Is 4 in squares?", 4 in squares)
print("Is 10 in squares?", 10 in squares)

# 3. Indexing and slicing
print("Second element:", squares[1])
print("Last element:", squares[-1])
print("From index 2:", squares[2:])

# 4. Concatenation and repetition
print("Concatenation:", squares + [36, 49, 64])
print("Repetition:", squares * 2)
print("Mixed types:", squares + ["one", 100, True])

# 5. Nested lists
a = ["a", "b", "c"]
n = [1, 2, 3]
x = [a, n]

print("Nested list:", x)
print("First list:", x[0])
print("Element b:", x[0][1])

# 6. Modify a list
cubes = [1, 8, 27, 65, 125]
cubes[3] = 64
print("After modification:", cubes)

# 7. List methods
cubes.append(216)
print("After append:", cubes)
print("Index of 216:", cubes.index(216))

cubes.reverse()
print("After reverse:", cubes)

cubes.sort()
print("After sort:", cubes)

cubes.remove(216)
print("After remove:", cubes)

# 8. Built-in functions
print("Minimum:", min(cubes))
print("Maximum:", max(cubes))
print("Sum:", sum(cubes))
```

```python


# Example 1: Both variables reference the same list
rgb = ["R", "G", "B"]
rgb_copy = rgb

rgb_copy.append("A")

print("Original rgb:", rgb)
print("rgb_copy:", rgb_copy)
print("Same object:", id(rgb) == id(rgb_copy))

print("-" * 40)

# Example 2: Create a separate copy using slicing
rgb = ["R", "G", "B"]
correct_rgb_copy = rgb[:]

correct_rgb_copy.append("A")

print("Original rgb:", rgb)
print("Correct copy:", correct_rgb_copy)
print("Same object:", id(rgb) == id(correct_rgb_copy))
```

Prime Numbers

```python
for n in range(10):

    if n < 2:
        continue


    for x in range(2, n):
        if n % x == 0:
            print(n, "equals", x, "*", n // x)
            break
    else:
        print(n, "is a prime number")
```

filter_even_numbers

```python

def filter_even_numbers(numbers):
    """
    Returns a list containing only even numbers.
    """

    result = []


    for number in numbers:


        if number % 2 == 0:
            result.append(number)

    return result

test = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = filter_even_numbers(test)


print("Original list:", test)
print("Even numbers:", even_numbers)
```

```python

# Exercise 1:


students = ["Alice", "Bob", "Charlie", "David", "Eve"]
grades = [85, 90, 78, 92, 90]

unique_grades = set(grades)
highest_grade = max(grades)

student_grades = dict(zip(students, grades))

top_students = [student for student, grade in student_grades.items() if grade == highest_grade]  #[cite: 2]

print("--- Exercise 1 Results ---")
print("Student Grades Dictionary:", student_grades)
print("Unique Grades Set:", unique_grades)
print("Highest Grade:", highest_grade)
print("Top Student(s):", top_students)



# Exercise 2:

def create_dict(keys, values):
    """
    Creates a dictionary from two lists if they have equal length.
    Returns an error message if lengths differ.
    """
    if len(keys) != len(values):
        return "Error: Lists must have the same length."
    return dict(zip(keys, values))


print("\n--- Exercise 2 Results ---")

keys_list = ["name", "age", "city"]
values_list = ["Alice", 25, "New York"]

result = create_dict(keys_list, values_list)
print("Generated Dictionary:", result)
```
